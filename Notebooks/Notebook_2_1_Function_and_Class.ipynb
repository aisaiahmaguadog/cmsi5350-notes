# Function

Syntax for functions with arguments:
```python
def function_name(parameter1: data_type1, parameter1: data_type2, ...) -> return_type:
    """Docstring"""
    # body of the function
    return value_to_return # Optional
```
Note that `data_type` and `return_type` are optional in function declaration, meaning the same function can also be written as:
```python
def function_name(parameter1, parameter2, ...):
    """Docstring"""
    # body of the function
    return value_to_return # Optional
```

As in loops and conditions, **you must indent in functions!**

Example:

```python
def sign(x: float) -> bool:
    if x >= 0:
        print('x is positive')
    elif x < 0:
        print('x is negative')
    else:
        pass # Tells the interpreter to do nothing

    return True
```

# Class

Classes are blueprints for creating objects in Python. They encapsulate data (attributes) and functionality (methods) into reusable structures, forming the foundation of object-oriented programming.

Basic class structure
```python
class ClassName:
    def __init__(self, parameters):
        self.attribute = value

    def method_name(self):
        return something
```
Key Components

* Constructor (`__init__`):
    * Special method that initializes new objects
    * Called automatically when creating an instance
    * Sets initial attribute values
* Attributes:
    * Variables that store object data
    * Instance attributes (unique to each object): `self.attribute`
* Methods:
    * Functions defined within a class
    * Always include self as first parameter
    * Access instance attributes and other methods
