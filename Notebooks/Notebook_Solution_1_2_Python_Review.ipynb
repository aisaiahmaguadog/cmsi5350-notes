### 1. **Temperature Converter**: Write code that converts temperatures between [Celsius and Fahrenheit](https://www.calculatorsoup.com/calculators/conversions/celsius-to-fahrenheit.php)

````python
#TODO
def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32

def fahrenheit_to_celsius(fahrenheit):
    return (fahrenheit - 32) * 5/9

temp_c = 25
temp_f = celsius_to_fahrenheit(temp_c)
print(f"{temp_c}°C = {temp_f}°F")
print(f"{temp_f}°F = {fahrenheit_to_celsius(temp_f):.1f}°C")
````

### 2. **Name Formatter**: Take a full name string and format it as "Last, First"

````python
#TODO
def format_name(full_name):
    parts = full_name.strip().split()
    if len(parts) >= 2:
        first_name = parts[0]
        last_name = " ".join(parts[1:])  # Handle multiple last names
        return f"{last_name}, {first_name}"
    return full_name

names = ["Alice Johnson", "Bob Smith Jr.", "Maria Garcia Lopez"]
for name in names:
    print(f"'{name}' -> '{format_name(name)}'")
````

## 🧩 Sets
Sets are unordered collections of **unique** elements. Handy for deduplicating labels, finding the vocabulary of a text corpus, or comparing which features two models used.

```python
all_labels = ["cat", "dog", "cat", "bird", "cat", "dog"]

unique_labels = set(all_labels)
print(unique_labels)
print(len(unique_labels))   # number of distinct classes

model_a_features = {"age", "income", "education"}
model_b_features = {"income", "education", "location"}

print(model_a_features & model_b_features)   # intersection
print(model_a_features | model_b_features)   # union
print(model_a_features - model_b_features)   # difference
```

````python
#TODO
all_labels = ["cat", "dog", "cat", "bird", "cat", "dog"]

unique_labels = set(all_labels)
print(unique_labels)
print(len(unique_labels))   # number of distinct classes

model_a_features = {"age", "income", "education"}
model_b_features = {"income", "education", "location"}

print(model_a_features & model_b_features)   # intersection
print(model_a_features | model_b_features)   # union
print(model_a_features - model_b_features)   # difference
````
